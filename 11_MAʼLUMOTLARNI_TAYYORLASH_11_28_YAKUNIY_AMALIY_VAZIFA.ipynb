#1) Quyida berilgan datasetni yuklab olib, ustun nomlarini koʻring va ularni maʼno jihatdan tushunarli shaklga keltiring. `Unnamed: 0` ustunini tashlab yuboring!

```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/alishermutalov/praktikum-datasets/refs/heads/praktikum/restaurant_sales.csv"
df = pd.read_csv(url)
df.head()
```

```python
#javobni shu yerda yozing
df = df.drop(columns=['Unnamed: 0'])
```

```python
df.rename(columns={'Order ID': 'OrderID', 'Customer ID': 'CustomerID', 'Order Total': 'TotalPrice', 'Order Date': 'Date', 'Payment Method':'PaymentMethod'}, inplace=True)
df.head()
```

#3) Har bir ustundagi boʻsh qiymatlarni aniqlang va ularni toʻldirish strategiyasini ishlab chiqing.

```python
#javobni shu yerda yozing
df.isnull().sum()
```

```python
# Item nomini ortiqcha belgilardan tozalash va ilk harflarini katta ko'rinishga keltirish
df['Item'] = df["Item"].str.extract("([a-zA-Z\s]+)", expand=False).str.strip().str.title()
df['Item']
```

```python
# categoriyani Itemga ko'ra aniqlash
item_category_map = (df.dropna(subset=['Item', 'Category']).drop_duplicates('Item').set_index('Item')['Category'].to_dict())
item_category_map
```

```python
df['Category'] = df.apply(lambda row: item_category_map.get(row['Item'], row['Category']), axis=1)
print(df['Category'].isna().sum())
```

```python
# qolga 4 ta Category nanlari qatorini o'chirib yuborish
df.drop(df[df['Category'].isna()].index, inplace=True)
```

```python
# Quantity nanlrini TotalPrice ni Price ga bo'lish va yana nan qolgan taqdirda min qiymat bilan to'ldirish
df['Quantity'] = df['Quantity'].fillna(df['TotalPrice'] / df['Price'])
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].min())

# TotalPrice da price va Quantity ni ko'paytirish orqali to'ldirish
df['TotalPrice'] = df['TotalPrice'].fillna(df['Price'] * df['Quantity']).abs()

# PaymentMethod nanlarini eng ko'p uchraydigan turi  bilan to'ldirish
df['PaymentMethod'] = df['PaymentMethod'].fillna(df['PaymentMethod'].mode().iloc[0])
```

```python
# Item ni kategoriyaga ko'ra aniqlash
category_item_map = (df.dropna(subset=['Category', 'Item']).groupby('Category')['Item'].agg(lambda x: x.mode()[0] if not x.mode().empty else 'Unknown').to_dict())
category_item_map
```

```python
# nan kategoriyalarni item categoriyasiga ko'ra to'ldirish
df['Item'] = df['Item'].fillna(df['Category'].map(category_item_map))
```

```python
# manfiy Price va TotalPrice larni musbatga o'zgartirish
df.loc[df['Price'] < 0, 'Price'] = df['Price'].abs()
df.loc[df['TotalPrice'] < 0, 'TotalPrice'] = df['TotalPrice'].abs()
```

```python
# narxni TotalPrice ga ko'ra hisoblash
df['Price'] = df['TotalPrice']/df['Quantity']
```
