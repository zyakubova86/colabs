 * https://en.wikipedia.org/wiki/Fossil_fuel_exporters
2. Har bir jadvalning birinchi 1 qatorini ekranga chiqaring.


    

```python
#javobni shu yerda yozing.
import pandas as pd
import numpy as np

tables = pd.read_html("https://en.wikipedia.org/wiki/Fossil_fuel_exporters")
for i, table in enumerate(tables):
  print(f"[Jadval {i+1}]")
  print(table.head(1))
```

#2-topshiriq
Yuqorida yaratilgan jadvallar roʻyxatidan eng uzunini `largest_data.csv` fayliga saqlang.

```python
#javobni shu yerda yozing.
longest_table = max(tables, key=lambda df: len(df))
longest_table.to_csv("largest_data.csv")
print("File saved")
```
